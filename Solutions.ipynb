# پاسخ پروژه‌های نهایی پایتون

این فایل شامل حل مستقل دو پروژه تحلیل فروش و تحلیل نمرات دانش‌آموزان است. سلول‌ها به ترتیب اجرا می‌شوند و نمودارها در خروجی نوت‌بوک نمایش داده می‌شوند.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## پروژه نهایی ۱: تحلیل فروش چندفروشه

```python
sales_df = pd.DataFrame({
    "seller": ["Ali", "Sara", "Ali", "Reza", "Sara", "Reza", "Ali", "Sara"],
    "product": ["Book", "Bag", "Pencil", "Notebook", "Backpack", "Pen", "Lamp", "Calculator"],
    "price": [120, 200, 30, 70, 180, 45, 90, 150],
    "quantity": [2, 1, 10, 3, 2, 5, 4, 1],
    "date": ["2024-01-01", "2024-01-01", "2024-01-02", "2024-01-02",
             "2024-01-03", "2024-01-03", "2024-01-04", "2024-01-04"]
})

sales_df["total_price"] = sales_df["price"] * sales_df["quantity"]
```

```python
# محاسبه فروش کل هر فروشنده
seller_sales = sales_df.groupby("seller")["total_price"].sum()
print("Total sales per seller:")
print(seller_sales)

# یافتن پرفروش‌ترین فروشنده
top_seller = seller_sales.idxmax()
print("Best-selling seller:", top_seller)

# نمودار ستونی فروش فروشندگان
seller_sales.plot(kind="bar")
plt.title("Total sales per seller")
plt.ylabel("Sales (thousand Tomans)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## پروژه نهایی ۲: تحلیل نمرات دانش‌آموزان

```python
grades_df = pd.DataFrame({
    "name": ["Ali", "Sara", "Reza", "Nilo", "Zahra", "Hasan", "Maryam", "Javad"],
    "class": ["A", "B", "A", "B", "A", "B", "A", "B"],
    "score": [18.5, 14.2, 11.8, 19.3, 16.7, 15.5, 10.4, 17.2]
})
```

```python
# محاسبه میانگین کلی
total_average = grades_df["score"].mean()
print("Overall class average:", total_average)

# محاسبه میانگین هر کلاس
average_per_class = grades_df.groupby("class")["score"].mean()
print("Average per class:", average_per_class)

# نمودار جعبه‌ای نمرات بر اساس کلاس
sns.boxplot(data=grades_df, x="class", y="score")
plt.title("Distribution of grades in different classes")
plt.tight_layout()
plt.show()

# نمودار پراکندگی نمرات بر اساس کلاس
sns.scatterplot(data=grades_df, x="name", y="score", hue="class")
plt.title("Student scores")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

🎯 <b>پروژه نهایی ۱: تحلیل فروش چندفروشه</b>

در این پروژه قصد داریم عملکرد فروشنده‌های مختلف یک فروشگاه را بررسی کنیم که   شامل اطلاعات زیر است:

- `seller`: نام فروشنده
- `product`: نام محصول
- `price`: قیمت واحد
- `quantity`: تعداد فروش
- `date`: تاریخ فروش

✅ مراحل تحلیل:
1. محاسبه فروش کل هر فروشنده
2. یافتن پرفروش‌ترین فروشنده
3. ترسیم نمودار ستونی از فروش فروشندگان
</div>

```python
import pandas as pd
import matplotlib.pyplot as plt

# ساخت داده فرضی برای فروش چندفروشه
df = pd.DataFrame({
    "seller": ["Ali", "Sara", "Ali", "Reza", "Sara", "Reza", "Ali", "Sara"],
    "product": ["Book", "Bag", "Pencil", "Notebook", "Backpack", "Pen", "Lamp", "Calculator"],
    "price": [120, 200, 30, 70, 180, 45, 90, 150],
    "quantity": [2, 1, 10, 3, 2, 5, 4, 1],
    "date": ["2024-01-01", "2024-01-01", "2024-01-02", "2024-01-02", "2024-01-03", "2024-01-03", "2024-01-04", "2024-01-04"]
})

# محاسبه ستون مبلغ فروش
df["total_price"] = df["price"] * df["quantity"]
```

<div dir="rtl">
<h3>💰 محاسبه فروش کل هر فروشنده (20 نمره)
</div>

```python
# فروش کل هر فروشنده
seller_sales = df.groupby("seller")["total_price"].sum()
print("💰 Total sales per seller:")
print(seller_sales)
```

<div dir="rtl">
<h3> 🏆 یافتن پرفروش‌ترین فروشنده
</div>

```python
# پرفروش‌ترین فروشنده
top_seller = seller_sales.idxmax()
print(f"🏆 Best-selling seller: {top_seller}")
```

<div dir="rtl">
<h3> 🔷 ترسیم نمودار ستونی از فروش فروشندگان (30 نمره)
</div>

```python
# نمودار ستونی
plot = seller_sales.plot(kind="bar")
plt.title("Total sales per seller")
plt.ylabel("Sales (thousand Tomans)")
plt.grid(True)
plt.show()
```

🎯 <b>پروژه نهایی ۲: تحلیل نمرات دانش‌آموزان</b>

در این پروژه داده‌های نمرات دانش‌آموزان بررسی می‌شود و نتایج به‌صورت تصویری نمایش داده خواهد شد.
 شامل ستون‌های:
- `name`: نام دانش‌آموز
- `class`: کلاس (A یا B)
- `score`: نمره نهایی

✅ مراحل تحلیل:
1. بررسی میانگین کلی و میانگین هر کلاس
2. نمودار جعبه‌ای نمرات بر اساس کلاس
3. نمودار پراکندگی نمرات بر اساس کلاس
</div>

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ساخت داده نمرات دانش‌آموزان
df = pd.DataFrame({
    "name": ["Ali", "Sara", "Reza", "Nilo", "Zahra", "Hasan", "Maryam", "Javad"],
    "class": ["A", "B", "A", "B", "A", "B", "A", "B"],
    "score": [18.5, 14.2, 11.8, 19.3, 16.7, 15.5, 10.4, 17.2]
})
```

<div dir="rtl">
<h3>🎓 نمایش میانگین کلی
</div>

```python
#محاسبه میانگین کلی
total_average = df["score"].mean()

# پرینت میانگین کلی
print("🎓 Overall class average:", total_average)
```

<div dir="rtl">
<h3>🎓 نمایش میانگین هر کلاس (20 نمره)
</div>

```python
# محاسبه میانگین هر کلاس
average_per_class = df.groupby("class")["score"].mean()

#پرینت میانگین هر کلاس
print("\n🎓 Average per class:", average_per_class)
```

<div dir="rtl">
<h3>🔷 نمودار جعبه‌ای نمرات بر اساس کلاس
</div>

```python
plot = sns.boxplot(data=df, x="class", y="score")
plt.title("Distribution of grades in different classes")
plt.show()
```

<div dir="rtl">
<h3>🔷 نمودار پراکندگی نمرات بر اساس کلاس (30 نمره)
</div>

```python
plot = sns.scatterplot(data=df, x="name", y="score", hue="class")
plt.title("Student scores")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```
